```python
import numpy as np
import pandas as pd
```

```python
np.random.seed(55)
```

```python
data = {
        "ID": np.arange (1,11),
        "Category": np.random.choice(['A','B','C'], size=10),
        "Value1": np.random.randint(10,100, size=10),
        "Value2": np.random.uniform(1.0,50.0, size=10)
}
df = pd.DataFrame(data)
print("Initial DataFrame:\n")
df
```

```python
df['Sum'] = np.add(df['Value1'], df['Value2']) # Value1 + Value2
df['Difference'] = np.subtract(df['Value1'], df['Value2']) # Value1 - Value2
df['SquareRoot_V1'] = np.sqrt(df['Value1']) # sqrt of Value1
df['Quotient'] = np.divide(df['Value1'], df['Value2'])
df['mean'] = np.mean(df['Value1'])
df['Product'] = np.multiply(df['Value1'], df['Value2'])

print('After Numpy Operations:\n')
df
```

```python
filtered_df = df[(df['Value1']>50) & (df['Category']=='A')] # Filtering DataFrame

print('Filtered DataFrame (Value1>50 7 Category == A): \n')
filtered_df
```

```python
grouped = df.groupby('Category').agg({ # grouping and aggregating
    'Value1': ['mean','max','min'],
    'Value2': ['mean', 'sum']
})

print('Grouped and aggregated data: \n')
grouped
```

```python
df["Value1_Normalized"] = df["Value1"].apply(lambda x: (x - np.min(df["Value1"])) / (np.max(df["Value1"]) - np.min(df["Value1"])))

print("With normalized Value1:\n")
df
```

```python
df.loc[2,'Value1'] = np.nan
df.loc[5,'Value2'] = np.nan

print('With NaN values:\n')
df
```
